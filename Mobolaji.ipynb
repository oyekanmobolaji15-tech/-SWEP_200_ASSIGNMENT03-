```python
class Swep():
  def __init__(self, uni, level, dept, count = 47): #number of students = 47
    self.uni = uni
    self.level = level
    self.dept = dept
    self.count = count

  def product_of_student_count_with_length_of_uni(self): #method, belongs to class
    product = self.count * self.level
    return product

uni = "OAU"
level = 250
dept = "CHE"

#STEP1: INITIALIZE THE CLASS
swep = Swep(uni, level, dept)

#step2: PERFORM THE ACTION: to perform an operation under a clas, use the dot notation, for example initialie class.method)
swep.product_of_student_count_with_length_of_uni()

```
